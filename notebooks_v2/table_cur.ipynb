```python
def extract_auc(val):
    return float(val.split(" ")[0])

auroc_cols = ["CLEAR", "CheXzero", "BiomedCLIP", "OpenAICLIP"]
# Function to format the best AUROC with LaTeX bold
def bold_max_latex(row):
    aucs = [extract_auc(row[col]) for col in auroc_cols]
    max_idx = aucs.index(max(aucs))
    formatted = {}
    for i, col in enumerate(auroc_cols):
        val = row[col]
        if i == max_idx:
            formatted[col] = f"\\textbf{{{val}}}"
        else:
            formatted[col] = val
    return pd.Series(formatted)

# Apply the bolding
df_latex = df_clip.sort_values(by='CLEAR', ascending=False).copy()
df_latex[auroc_cols] = df_clip.apply(lambda row: pd.Series(bold_max_latex(row), index=auroc_cols), axis=1)
# df_latex = df_latex.sort_values(by='CLEAR', ascending=False)
# Export to LaTeX code string
latex_code = df_latex.to_latex(index=False, escape=False, column_format="llcccc")

# Print or save
print(latex_code)
```

```python
df_clear = pd.read_csv('/home/than/DeepLearning/cxr_concept/CheXzero/concepts/results/linear_probing_auc_comparison/sfr_mistral_detailed_results.csv')
df_chexzero = pd.read_csv('/home/than/DeepLearning/cxr_concept/CheXzero/concepts/results/linear_probing_auc_comparison/chexzero_detailed_results.csv')
df_biomedclip = pd.read_csv('/home/than/DeepLearning/cxr_concept/CheXzero/concepts/results/linear_probing_auc_comparison/biomedclip_detailed_results.csv')
df_openaiclip = pd.read_csv('/home/than/DeepLearning/cxr_concept/CheXzero/concepts/results/linear_probing_auc_comparison/openai_clip_detailed_results.csv')
df_openaiclip.head()
```

```python
def format_df(df, model_name, auc_col, lower_col, upper_col):
    return df[["label", auc_col, lower_col, upper_col]].copy().rename(columns={
        auc_col: f"{model_name}_auc",
        lower_col: f"{model_name}_ci_lower",
        upper_col: f"{model_name}_ci_upper"
    })

clear = format_df(df_clear, "CLEAR", "sfr_mistral_mean_auc", "sfr_mistral_ci_lower", "sfr_mistral_ci_upper")
chexzero = format_df(df_chexzero, "CheXzero", "chexzero_mean_auc", "chexzero_ci_lower", "chexzero_ci_upper")
biomedclip = format_df(df_biomedclip, "BiomedCLIP", "biomedclip_mean_auc", "biomedclip_ci_lower", "biomedclip_ci_upper")
openai = format_df(df_openaiclip, "OpenAICLIP", "openai_clip_mean_auc", "openai_clip_ci_lower", "openai_clip_ci_upper")

# Merge all on label
df = clear.merge(chexzero, on="label").merge(biomedclip, on="label").merge(openai, on="label")

# Define model column groups
model_names = ["CLEAR", "CheXzero", "BiomedCLIP", "OpenAICLIP"]

# Format AUROC + CI and bold the max per row
def row_formatter(row):
    aucs = []
    for m in model_names:
        auc = row[f"{m}_auc"]
        low = row[f"{m}_ci_lower"]
        high = row[f"{m}_ci_upper"]
        aucs.append((m, auc, f"{auc:.3f} ({low:.3f}, {high:.3f})"))
    # Find the model with max AUROC
    max_model = max(aucs, key=lambda x: x[1])[0]
    # Create output row with bolding
    output = {}
    for m, auc, text in aucs:
        output[m] = f"\\textbf{{{text}}}" if m == max_model else text
    return pd.Series(output)

# Apply formatting to the merged df
formatted_df = df[["label"]].copy()
formatted_df[model_names] = df.apply(row_formatter, axis=1)
```

```python
latex_code = formatted_df.to_latex(index=False, escape=False, column_format="lcccc")

# Print or save
print(latex_code)
```
